```python
import nltk
from nltk import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
import numpy as np
import random
import pickle
import json

from keras import Sequential
from keras.layers import Dense, Dropout
from keras.losses import SparseCategoricalCrossentropy

```

```python
words = []
classes = []
documents = []
ignor_ltter = [',','.','?','!']
intent_fil = open('intents.json').read()
intents = json.loads(intent_fil)

for intent in intents['intents']:
    for pattern in intent['patterns']:
        word = nltk.wordpunct_tokenize(pattern)
        words.extend(word)
        documents.append((word,intent['tag']))
        if intent['tag'] not in classes:
            classes.append(intent['tag'])

print(documents)
```

```python
words = [lemmatizer.lemmatize(w.lower())for w in words if w not in ignor_ltter]
words = sorted(list(set(words)))
classes = sorted(list(set(classes)))

print('documents',len(documents))
print('words',len(words),words)
print('classes',len(classes),classes)
```

```python
training = []
output_empty = [0]*len(classes)
for doc in documents:
    bag = []
    pattern_words = doc[0]
    for word in words:
        bag.append(1) if word in pattern_words else bag.append(0)
    output = list(output_empty)
    output[classes.index(doc[1])]=1
    training.append([bag,output])
random.shuffle(training)
training = np.array(training,dtype=object)
train_x = list(training[:,0])
train_y = list(training[:,1])
print(training)
    
```

```python
model = Sequential()
model.add(Dense(128, input_shape=(len(train_x[0]),), activation='relu'))
model.add(Dropout(0.5))
model.add(Dense(64, activation='relu'))
model.add(Dropout(0.5))
model.add(Dense(len(train_y[0]), activation='softmax'))
model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
hist = model.fit(np.array(train_x), np.array(train_y), epochs=200, batch_size=5, verbose=1)
model.save('chatbot_model.h5', hist)
print("model created")
```
